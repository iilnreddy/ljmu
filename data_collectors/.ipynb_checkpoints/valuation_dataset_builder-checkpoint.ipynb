# Historical Valuation Dataset Builder (S&P500 + NIFTY500)
This notebook downloads price and fundamental data using yfinance and computes valuation ratios such as PE, PB, PS, ROE and revenue growth.

Dataset pipeline:
1. Load stock universe
2. Download historical prices
3. Download financial statements
4. Merge datasets
5. Compute valuation factors
6. Export dataset

```python
import pandas as pd
import yfinance as yf
from tqdm import tqdm
```

## Step 1 — Load S&P500 tickers

```python
sp500 = pd.read_html('https://en.wikipedia.org/wiki/List_of_S%26P_500_companies')[0]
tickers_sp500 = sp500['Symbol'].tolist()
tickers_sp500[:10]
```

## Step 2 — Load NIFTY500 tickers
Place a file named `nifty500.csv` with a column called `Symbol`.

```python
nifty = pd.read_csv('nifty500.csv')
tickers_nifty = [s + '.NS' for s in nifty['Symbol'].tolist()]
tickers_nifty[:10]
```

## Step 3 — Functions to download price and fundamentals

```python
def get_price(ticker):
    df = yf.download(ticker, start='2010-01-01', progress=False)
    df = df.reset_index()
    df['ticker'] = ticker
    return df
```

```python
def get_fundamentals(ticker):
    t = yf.Ticker(ticker)
    income = t.quarterly_income_stmt.T
    balance = t.quarterly_balance_sheet.T

    df = pd.DataFrame()
    
    if 'Diluted EPS' in income.columns:
        df['eps'] = income['Diluted EPS']
    
    if 'Total Revenue' in income.columns:
        df['revenue'] = income['Total Revenue']
        
    if 'Net Income' in income.columns:
        df['net_income'] = income['Net Income']

    if 'Total Stockholder Equity' in balance.columns:
        df['equity'] = balance['Total Stockholder Equity']

    df['date'] = df.index
    df['ticker'] = ticker

    return df.reset_index(drop=True)
```

## Step 4 — Merge price and fundamentals

```python
def merge_data(price, fundamentals):
    fundamentals['date'] = pd.to_datetime(fundamentals['date'])

    merged = pd.merge_asof(
        price.sort_values('Date'),
        fundamentals.sort_values('date'),
        left_on='Date',
        right_on='date',
        direction='backward'
    )

    return merged
```

## Step 5 — Compute valuation factors

```python
def compute_factors(df):

    df['PE'] = df['Close'] / df['eps']

    df['PB'] = df['Close'] / (df['equity'] / 1e9)

    df['PS'] = df['Close'] / (df['revenue'] / 1e9)

    df['ROE'] = df['net_income'] / df['equity']

    df['revenue_growth'] = df['revenue'].pct_change(4)

    return df
```

## Step 6 — Build dataset

```python
all_tickers = tickers_sp500 + tickers_nifty

dataset = []

for ticker in tqdm(all_tickers):

    try:
        price = get_price(ticker)
        fundamentals = get_fundamentals(ticker)

        merged = merge_data(price, fundamentals)
        factors = compute_factors(merged)

        dataset.append(factors)

    except Exception as e:
        print('Error:', ticker, e)
```

## Step 7 — Final dataset

```python
final_df = pd.concat(dataset)

final_df.head()
```

## Step 8 — Save dataset

```python
final_df.to_parquet('valuation_dataset.parquet')
final_df.to_csv('valuation_dataset.csv', index=False)

print('Dataset saved')
```
